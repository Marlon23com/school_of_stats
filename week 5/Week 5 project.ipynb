Reading the entire file into a variable.

```python
# Open and read the file
with open('week5_customer_reviews.csv', 'r', encoding='utf-8') as f:
    raw_data = f.read()
```

Print out the first few lines so we can see what we’re working with.

```python
# Print first 500 characters
print(raw_data[:500])
```

Because the file is not properly formatted, will split it into rows manually.

```python
# Split the data into rows using commas followed by newlines
rows = raw_data.split(',,')
```

```python
import re

cleaned_rows = []

for row in rows:
    # Remove leading/trailing whitespace
    row = row.strip()
    
    # Skip empty rows
    if not row:
        continue

    # Split by commas, but not inside quoted strings (basic version)
    parts = re.split(r',(?=(?:[^"]*"[^"]*")*[^"]*$)', row)

    # Store cleaned row
    cleaned_rows.append(parts)
```
