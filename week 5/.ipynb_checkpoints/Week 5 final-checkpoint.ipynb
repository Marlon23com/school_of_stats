Question 4: How can you remove all special characters, emojis, and numbers from the raw_review column using Python?

```python
# Use regular expressions

import re

def clean_review(text):
    # Remove emojis
    emoji_pattern = re.compile(
        "["
        u"\U0001F600-\U0001F64F"  # emoticons
        u"\U0001F300-\U0001F5FF"  # symbols & pictographs
        u"\U0001F680-\U0001F6FF"  # transport & map symbols
        u"\U0001F1E0-\U0001F1FF"  # flags (iOS)
        "]+", flags=re.UNICODE)
    
    text = emoji_pattern.sub(r'', text)
    
    # Remove special characters and digits
    text = re.sub(r'[^a-zA-Z\s]', '', text)
    
    # Remove extra spaces
    text = re.sub(r'\s+', ' ', text).strip()
    
    return text
```

 Question 5: Write a function in Python that strips extra spaces and lowercases the review text.

```python
def preprocess_text(text):
    text = text.lower().strip()
    return text

# Apply both functions
df['cleaned_review'] = df['raw_review'].apply(clean_review).apply(preprocess_text)
```

Question 7: How would you compute the number of words in each review and store it in a new column review_length?

```python
df['review_length'] = df['cleaned_review'].apply(lambda x: len(x.split()))
```

Question 8: Can you categorize reviews into ‘Positive’, ‘Neutral’, or ‘Negative’ using the rating column? Create a new column sentiment.

```python
def get_sentiment(rating):
    if rating >= 4:
        return 'Positive'
    elif rating == 3:
        return 'Neutral'
    else:
        return 'Negative'

df['sentiment'] = df['rating'].apply(get_sentiment)
```

Question 9: Which product received the most negative reviews (rating ≤ 2)?

```python
negative_reviews = df[df['rating'] <= 2]
product_counts = negative_reviews['product'].value_counts()

most_negative_product = product_counts.idxmax()
print(f"Product with most negative reviews: {most_negative_product}")
print(product_counts)
```

 Question 10: What’s the average review length for positive vs negative reviews?

```python
avg_lengths = df.groupby('sentiment')['review_length'].mean()
print(avg_lengths)
```

Question 11: Plot a bar chart showing the number of reviews per product. What insights can you draw?

```python
import matplotlib.pyplot as plt

product_counts = df['product'].value_counts()
product_counts.plot(kind='bar', figsize=(10, 6), color='skyblue')
plt.title('Number of Reviews per Product')
plt.xlabel('Product')
plt.ylabel('Number of Reviews')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

Question 12: Do longer reviews tend to be more positive or more negative based on the review_length and rating relationship?

```python
import seaborn as sns

sns.boxplot(x='sentiment', y='review_length', data=df, order=['Negative', 'Neutral', 'Positive'])
plt.title('Review Length by Sentiment')
plt.xlabel('Sentiment')
plt.ylabel('Review Length')
plt.show()
```
