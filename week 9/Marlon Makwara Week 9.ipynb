```python
# Import libraries
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

# Set plot styles
sns.set(style="whitegrid")
plt.style.use("seaborn-v0_8-darkgrid")

# Load the dataset
df = pd.read_csv('week9_titanic_eda_data.csv')

# Preview the dataset
df.head()
```

```python
# Number of rows and columns
df.shape
```

The dataset contains 5000 passengers and 12 columns

```python
# Dataset information
df.info()

# Summary statistics
df.describe(include='all')
```

```python
# Count of survivors and non-survivors
survival_counts = df['Survived'].value_counts()
print(survival_counts)

# Bar chart
plt.figure(figsize=(6,4))
sns.countplot(x='Survived', data=df, palette='Set2')
plt.title('Survival Distribution')
plt.xlabel('Survived (0 = No, 1 = Yes)')
plt.ylabel('Passenger Count')
plt.xticks([0, 1], ['Did Not Survive', 'Survived'])
plt.show()
```

This shows 2496 passengers survived vs 2504 who did not survive

```python
# Count of passenger classes
pclass_counts = df['Pclass'].value_counts()
print(pclass_counts)

# Bar chart
plt.figure(figsize=(6,4))
sns.countplot(x='Pclass', data=df, palette='pastel')
plt.title('Passenger Class Distribution')
plt.xlabel('Passenger Class')
plt.ylabel('Count')
plt.show()
```

```python
# Pie chart
plt.figure(figsize=(6,6))
df['Pclass'].value_counts().plot.pie(autopct='%1.1f%%', startangle=90, colors=['#66b3ff','#99ff99','#ffcc99'])
plt.title('Passenger Class Distribution')
plt.ylabel('')
plt.show()
```

```python
# Check missing values in Age
missing_age = df['Age'].isnull().sum()
print(f"Missing Age values: {missing_age}")

# Histogram of Age
plt.figure(figsize=(8,4))
sns.histplot(df['Age'].dropna(), bins=30, kde=True, color='skyblue')
plt.title('Age Distribution of Passengers')
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.boxplot(x='Survived', y='Age', data=df, palette='Set2')
plt.title('Age vs Survival')
plt.xlabel('Survived (0 = No, 1 = Yes)')
plt.ylabel('Age')
plt.xticks([0, 1], ['Did Not Survive', 'Survived'])
plt.show()
```

```python
# Count of each gender
gender_counts = df['Sex'].value_counts()
print(gender_counts)

# Bar chart
plt.figure(figsize=(6,4))
sns.countplot(x='Sex', data=df, palette='coolwarm')
plt.title('Gender Distribution')
plt.xlabel('Gender')
plt.ylabel('Count')
plt.show()
```

```python
# Cross-tabulation of gender and survival
gender_survival = pd.crosstab(df['Sex'], df['Survived'])
print(gender_survival)

# Stacked bar chart
gender_survival.plot(kind='bar', stacked=True, colormap='coolwarm', figsize=(6,4))
plt.title('Survival by Gender')
plt.xlabel('Gender')
plt.ylabel('Count')
plt.legend(['Did Not Survive', 'Survived'], loc='upper right')
plt.show()
```

```python
# Summary statistics
print(df[['SibSp', 'Parch']].describe())

# Histograms
fig, axes = plt.subplots(1, 2, figsize=(12,4))

sns.histplot(df['SibSp'], bins=8, ax=axes[0], color='lightcoral')
axes[0].set_title('Siblings/Spouses Aboard (SibSp)')
axes[0].set_xlabel('SibSp')

sns.histplot(df['Parch'], bins=8, ax=axes[1], color='lightblue')
axes[1].set_title('Parents/Children Aboard (Parch)')
axes[1].set_xlabel('Parch')

plt.tight_layout()
plt.show()
```

```python
# Count of embarkation ports
embarked_counts = df['Embarked'].value_counts()
print(embarked_counts)

# Bar chart
plt.figure(figsize=(6,4))
sns.countplot(x='Embarked', data=df, palette='Set3')
plt.title('Embarkation Port Distribution')
plt.xlabel('Port (C = Cherbourg, Q = Queenstown, S = Southampton)')
plt.ylabel('Count')
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.boxplot(x='Pclass', y='Fare', data=df, palette='pastel')
plt.title('Fare Distribution by Passenger Class')
plt.xlabel('Passenger Class')
plt.ylabel('Fare')
plt.show()
```

Is there any relationship between Fare and Survival?

```python
# Boxplot: Fare vs Survival
plt.figure(figsize=(8,5))
sns.boxplot(x='Survived', y='Fare', data=df, palette='coolwarm')
plt.title('Fare vs Survival')
plt.xlabel('Survived (0 = No, 1 = Yes)')
plt.ylabel('Fare')
plt.show()

# Optional scatterplot (with jitter)
plt.figure(figsize=(8,5))
sns.stripplot(x='Survived', y='Fare', data=df, jitter=True, palette='coolwarm', alpha=0.5)
plt.title('Scatter of Fare by Survival')
plt.xlabel('Survived (0 = No, 1 = Yes)')
plt.ylabel('Fare')
plt.show()
```

How many missing values are there in each column? Visualize with heatmap

```python
# Missing values
missing_values = df.isnull().sum()
print(missing_values)

# Heatmap
plt.figure(figsize=(10,6))
sns.heatmap(df.isnull(), cbar=False, cmap='viridis', yticklabels=False)
plt.title('Missing Values Heatmap')
plt.show()
```

```python
# Group by survival status and calculate mean age
avg_ages = df.groupby('Survived')['Age'].mean()
print(avg_ages)
```

correlation heatmap for numerical variables

```python
# Correlation matrix
corr_matrix = df.corr(numeric_only=True)

# Heatmap
plt.figure(figsize=(10,6))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', linewidths=0.5)
plt.title('Correlation Heatmap')
plt.show()
```

stacked bar chart showing survival by passenger class

```python
# Cross-tabulation
pclass_survival = pd.crosstab(df['Pclass'], df['Survived'])
print(pclass_survival)

# Stacked bar chart
pclass_survival.plot(kind='bar', stacked=True, colormap='viridis', figsize=(6,4))
plt.title('Survival by Passenger Class')
plt.xlabel('Passenger Class')
plt.ylabel('Count')
plt.legend(['Did Not Survive', 'Survived'])
plt.show()
```
